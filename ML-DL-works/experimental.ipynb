```python

import cv2
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# 1. Load your image
image = cv2.imread("Untitled.jpeg")

# Check if image was loaded
if image is None:
    print("Error: Image not found!")
    exit()

# OpenCV loads images as BGR
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# 2. Get image dimensions
height, width, channels = image_rgb.shape

# 3. Convert image into a 2D array of pixels
# Each pixel becomes [R, G, B]
pixels = image_rgb.reshape(-1, 3)

print("Image shape:", image_rgb.shape)
print("Number of pixels:", len(pixels))

# 4. Choose number of clusters
k = 2

# 5. Create K-Means model
kmeans = KMeans(
    n_clusters=k,
    random_state=42,
    n_init=10
)

# 6. Fit K-Means
kmeans.fit(pixels)

# 7. Get cluster labels
labels = kmeans.labels_

# 8. Get cluster centers/colors
colors = kmeans.cluster_centers_.astype(np.uint8)

print("Cluster colors:")
print(colors)

# 9. Replace every pixel with its cluster center color
clustered_pixels = colors[labels]

# 10. Convert back to image shape
clustered_image = clustered_pixels.reshape(height, width, 3)

# 11. Display original and clustered image
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(image_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(clustered_image)
plt.title(f"K-Means Clustering (K={k})")
plt.axis("off")

plt.show()
```
