```python
# Noise Removal and Normalization

import string

text = "Hello, World! This is a sample text with punctuation and mixed CASE."

# 1. LOWERCASE CONVERSION
text_lower = text.lower()

# 2. Remove punctuation using string translation
translator = str.maketrans('', '', string.punctuation)
clean_text = text_lower.translate(translator)

print("Original Text:", text)
print("Clean Text:", clean_text)
```

```python
# Tokenization
tokens = clean_text.split()
print("Tokens:", tokens)
print("Number of Tokens:", len(tokens))
```

```python
# Subword Tokenization BERT
inputs = " The Microtransactional system was counterintuitive"
from transformers import AutoTokenizer
tokenizer = AutoTokenizer.from_pretrained('bert-base-uncased')
inputs = tokenizer.tokenize(inputs)
print("BERT Tokenization:", inputs)
```

```python
# Stopword Removal
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')

stop_words = set(stopwords.words('english'))
filtered_tokens = [word for word in tokens if word not in stop_words]
print("Filtered Tokens:", filtered_tokens)
print("Number of Filtered Tokens:", len(filtered_tokens))
print("Stopwords Removed:", set(tokens) - set(filtered_tokens))
```

```python
# Stemming & Lemmatization
from nltk.stem import PorterStemmer, WordNetLemmatizer
nltk.download('wordnet')
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

for word in filtered_tokens:

    print(word, "|", stemmer.stem(word), "|", lemmatizer.lemmatize(word))
```

```python
# Vector Embeddings
from sentence_transformers import SentenceTransformer
model = SentenceTransformer('all-MiniLM-L6-v2')
words = ["King", "Queen", "Man", "Woman"]
embeddings = model.encode(words)

for i in range(3):
    print("Words : ", words[i])
    print("Embeddings : ", embeddings[i][:3])  # Print first 3 dimensions of the embedding

# cosine similarity
similarity = model.similarity([embeddings[0]], [embeddings[2]])
print("Cosine Similarity between 'King' and 'Queen':", similarity)
```
